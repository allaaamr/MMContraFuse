```python
import pandas as pd

cna = pd.read_csv("combined_CNA_filtered.csv")
rna = pd.read_csv("combined_RNA_filtered.csv")
mut = pd.read_csv("combined_MUT_filtered.csv")
clinical = pd.read_csv("clinical.csv")
```

```python
cna.shape, rna.shape, clinical.shape
```

```python
cna_clinical = pd.merge(clinical,cna, left_on="case_id", right_on="PatientID", how="inner")
# Custom Dataset Expects slide id of pathology image in a column... for now we will add it as a dummy entry untill we do the pathology
cna_clinical.insert(
    loc=1,                # index position (0 = first col, 1 = after first col)
    column="slide_id",       # new column name
    value="x"              # same dummy value for all rows
)

cna_clinical.isna().sum().sort_values(ascending=False)
```

```python
cna_clinical = cna_clinical.dropna(subset=['survival'])
cna_clinical.to_csv("cna_clinical.csv")
cna_clinical.shape, cna_clinical.head()
```

```python
rna_clinical = pd.merge(clinical,rna, left_on="case_id", right_on="PatientID", how="inner")
# Custom Dataset Expects slide id of pathology image in a column... for now we will add it as a dummy entry untill we do the pathology
rna_clinical.insert(
    loc=1,                # index position (0 = first col, 1 = after first col)
    column="slide_id",       # new column name
    value="x"              # same dummy value for all rows
)
rna_clinical.isna().sum().sort_values(ascending=False)
# rna_clinical.to_csv("rna_clinical.csv")
# rna_clinical.shape, rna_clinical.head()
```
